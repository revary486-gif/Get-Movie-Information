# 260809 씨네21 홈페이지 스크랩

```python
import requests
from bs4 import BeautifulSoup
import sys

print(sys.executable)
```

```python
url = "https://cine21.com/"
header = {
    "User-Agent": "Mozilla/5.0 (compatible; EducationalScraper/1.0; +https://example.com)"
}

response = requests.get(url, headers = header)

print("상태 코드: ", response.status_code)
```

```python
html = response.text
```

```python
soup = BeautifulSoup(html, "html.parser")
```

```python
movie_rank = []
movie_title = []
movie_stars = []
movie_weekly_viewer = []
movie_cumulative_viewer = []
movie_detail_pages = []
movie_release_date = []
etc_infos = []

# 클래스 이름들
# div 태그 내 사용해야 할 클래스(rank, etc_info 텍스트 3개씩 나오는 것)
# p 태그 내 사용될 클래스(title, num)
div_class = ['rank', 'etc_info']
```

```python
items = soup.select("div.list_with_upthumb_item div.rank")
items_2 = soup.select("div.list_with_upthumb_item div.etc_info p")
items_3 = soup.select("div.swiper-wrapper div.list_with_upthumb_item p.title")
items_4 = soup.select("div.list_with_upthumb_item p.num")
```

```python
for i in items:
    rank = i.get_text()
    movie_rank.append(rank)
for j in items_2:
    for p in soup.select("div.list_with_upthumb_item div.etc_info p"):
            etc_info = [p.get_text()]
            etc_infos.append(etc_info)
for k in items_3:
    title = k.get_text()
    movie_title.append(title)
for q in items_4:
    star = q.get_text()
    movie_stars.append(star)

print(movie_rank)
print(etc_infos[8:38])
total_etc_info = etc_infos[8:38]
print(movie_title[5:15])
movie_titles = movie_title[5:15]
print(movie_stars)
```

```python
movie_weekly_viewer = total_etc_info[0::3]
movie_cumulative_viewer = total_etc_info[1::3]
for i in range(0, 10):
     movie_cumulative_viewer[i][0] = movie_cumulative_viewer[i][0].replace('누적 : ', '')
movie_release_date = total_etc_info[2::3]
for k in range(0, 10):
     movie_release_date[k][0] = movie_release_date[k][0].replace('개봉일 : ', '')

movie_cumulative_viewer = [x[0] for x in movie_cumulative_viewer]
movie_weekly_viewer = [x[0] for x in movie_weekly_viewer]
movie_release_date = [x[0] for x in movie_release_date]

print(movie_weekly_viewer)
print(movie_cumulative_viewer)
print(movie_release_date)
```

```python
movie_detail_page_links = []
links = soup.select('div.list_with_upthumb_item>a[href^="/movie/info"]')

for link in links:
    href = link.get("href")
    movie_detail_page_links.append(href)
print(len(movie_detail_page_links))
```

```python
from openpyxl import Workbook
```

```python
wb = Workbook(write_only=True)
ws = wb.create_sheet()
ws.append(['제목', '순위', '개봉일', '평점', '주간_관객수', '누적_관객 수', '상세페이지'])

for row in zip(movie_titles, movie_rank, movie_release_date, movie_stars, movie_weekly_viewer, movie_cumulative_viewer, movie_detail_page_links):
    ws.append(row)

wb.save('260809_cine21_main_page_final.xlsx')
```
