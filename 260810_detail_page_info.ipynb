### 라이브러리 임포트

```python
import requests
from bs4 import BeautifulSoup
from openpyxl import Workbook
import sys

print(sys.executable)
```

```python
url = "https://cine21.com/movie/info/?movie_id=63091"

header = {
    "User-Agent": "Mozilla/5.0 (compatible; EducationalScraper/1.0; +https://example.com)"
}
response = requests.get(url, headers = header)

print("상태코드 : ", response.status_code)
```

```python
html = response.text
soup = BeautifulSoup(html, "html.parser")
```

```python
# 개봉일 / 등급 / 시간 / 누적 관객수 / 장르 / 국가 정보
info_list = []
items = soup.select("section.wide_dark_bg_wrap ul.info_list li p.title")
for i in items:
    n = i.next_sibling
    info = n.get_text(strip=True)
    info_list.append(info)

print(info_list)
```

```python
# 영화 감독 / 출연진
info_list2 = []
items = soup.select("section.wide_dark_bg_wrap ul.info_list a")
for i in items:
    info = i.get_text(strip=True)
    info_list2.append(info)

print(info_list2)
```

```python
# 영화 아이디 리스트
movie_id_list = [63091, 62585, 63075, 62480, 63032, 62799, 63282, 62818, 63108, 63240]
```

```python
info_list = []
temp_list = []
info_list2 = []

for d in movie_id_list:
    url = f'https://cine21.com/movie/info/?movie_id={d}'
    response = requests.get(url, headers = header)
    html = response.text
    soup = BeautifulSoup(html, "html.parser")
    items = soup.select("section.wide_dark_bg_wrap ul.info_list li p.title")
    temp_list = []
    for i in items:
        n = i.next_sibling
        info = n.get_text(strip=True)
        temp_list.append(info)

    info_list.append(temp_list)


print(info_list)
```

```python
release_dates = []
movie_ratings = []
running_times = []
cumulative_viewers = []
movie_genres = []
release_countries =[]

for i in range(0, 10):
    date = info_list[i][0]
    release_dates.append(date)
    rating = info_list[i][1]
    movie_ratings.append(rating)
    running_time = info_list[i][2]
    running_times.append(running_time)
    cumulative_viewer = info_list[i][3]
    cumulative_viewers.append(cumulative_viewer)
    genre = info_list[i][4]
    movie_genres.append(genre)
    country = info_list[i][5]  
    release_countries.append(country)
```

```python
print(release_dates)
print(movie_ratings)
```

```python
info_list2 = []

for d in movie_id_list:
    url = f'https://cine21.com/movie/info/?movie_id={d}'
    response = requests.get(url, headers = header)
    html = response.text
    soup = BeautifulSoup(html, "html.parser")
    items = soup.select("section.wide_dark_bg_wrap ul.info_list a")
    temp2_list = []
    for i in items:
        info = i.get_text(strip=True)
        temp2_list.append(info)
    
    info_list2.append(temp2_list)

print(info_list2)
```

```python
movie_directors = []

for k in range(0, 10):
    director = info_list2[k][0]
    movie_directors.append(director)

print(movie_directors)
```

```python
wb = Workbook(write_only=True)
ws = wb.create_sheet()
ws.append(['개봉일', '등급', '상영시간', '누적_관객수', '장르', '국가', '감독'])

for row in zip(release_dates, movie_ratings, running_times, cumulative_viewers, movie_genres, release_countries, movie_directors):
    ws.append(row)

wb.save('260809_cine21_detail_page.xlsx')
```

```python
import csv
csv_filename = "260809_cine21_detail_page.csv"

with open(csv_filename, mode="w", newline="", encoding="utf-8-sig") as file:
    writer = csv.writer(file, delimiter=",")
    writer.writerow(
        [
            "개봉일",
            "등급",
            "상영시간",
            "누적_관객수",
            "장르",
            "국가",
            "감독",
        ]
    )

    for row in zip(
        release_dates,
        movie_ratings,
        running_times,
        cumulative_viewers,
        movie_genres,
        release_countries,
        movie_directors,
    ):
        writer.writerow(row)
```

## 전문가 평가
- 태그 및 클래스가 두 개로 나뉘어져 있어 각각 확인
- 정상적으로 출력되는 것 확인 후 합쳐서 진행

```python
expert_evaluations = []

for id in movie_id_list:
    url = f'https://cine21.com/movie/info/?movie_id={id}'
    response = requests.get(url, headers = header)
    html = response.text
    soup = BeautifulSoup(html, "html.parser")
    items = soup.select("section.expert_star div.reviewer p")
    temp_list3 = []
    for i in items:
        num = i.get_text()
        temp_list3.append(num)
    for i in range(0, len(temp_list3), 2):
        expert_evaluations.append({
            "movie_id" : id,
            "expert_name" : temp_list3[i],
            "expert_star" : temp_list3[i+1]
        })

print(expert_evaluations)
print(len(expert_evaluations))
```

```python
expert_reviews = []

for id in movie_id_list:
    url = f'https://cine21.com/movie/info/?movie_id={id}'
    response = requests.get(url, headers = header)
    html = response.text
    soup = BeautifulSoup(html, "html.parser")
    items = soup.select("section.expert_star div.review")
    review_temp_list = []
    for i in items:
        review = i.get_text(strip=True)
        review_temp_list.append(review)
    expert_reviews.append(review_temp_list)
print(expert_reviews)
```

```python
expert_evaluations = []

for id in movie_id_list:
    url = f'https://cine21.com/movie/info/?movie_id={id}'
    response = requests.get(url, headers = header)
    html = response.text
    soup = BeautifulSoup(html, "html.parser")
    items = soup.select("section.expert_star div.reviewer p")
    temp_list3 = []
    for i in items:
        num = i.get_text()
        temp_list3.append(num)
    url = f'https://cine21.com/movie/info/?movie_id={id}'
    response = requests.get(url, headers = header)
    html = response.text
    soup = BeautifulSoup(html, "html.parser")
    items = soup.select("section.expert_star div.review")
    review_temp_list = []
    for i in items:
        review = i.get_text(strip=True)
        review_temp_list.append(review)

    for i in range(0, len(temp_list3), 2):
            expert_evaluations.append({
                "movie_id" : id,
                "expert_name" : temp_list3[i],
                "expert_star" : temp_list3[i+1],
                "expert_review" : review_temp_list[i // 2]
            })

print(expert_evaluations)
print(len(expert_evaluations))
```

### 엑셀 및 csv 파일로 저장

```python
wb = Workbook(write_only=True)
ws = wb.create_sheet()

ws.append(["movie_id", "전문가_이름", "전문가_평점", "전문가_리뷰"])

for data in expert_evaluations:
    ws.append([
        data["movie_id"],
        data["expert_name"],
        data["expert_star"],
        data["expert_review"]
    ])

wb.save("260809_movie_expert_review.xlsx")
```

```python
csv_filename = "260809_movie_expert_review.csv"

with open(csv_filename, mode="w", newline="", encoding="utf-8-sig") as file:
    writer = csv.writer(file, delimiter=",")
    writer.writerow(["movie_id", "전문가_이름", "전문가_평점", "전문가_리뷰"])

    for data in expert_evaluations:
        writer.writerow([
            data["movie_id"],
            data["expert_name"],
            data["expert_star"],
            data["expert_review"]
        ])
```
