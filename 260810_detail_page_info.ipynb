```python
import requests
from bs4 import BeautifulSoup
from openpyxl import Workbook
import sys

print(sys.executable)
```

```python
url = "https://cine21.com/movie/info/?movie_id=63091"

header = {
    "User-Agent": "Mozilla/5.0 (compatible; EducationalScraper/1.0; +https://example.com)"
}
response = requests.get(url, headers = header)

print("상태코드 : ", response.status_code)
```

```python
html = response.text
soup = BeautifulSoup(html, "html.parser")
```

```python
# 개봉일 / 등급 / 시간 / 누적 관객수 / 장르 / 국가 정보
info_list = []
items = soup.select("section.wide_dark_bg_wrap ul.info_list li p.title")
for i in items:
    n = i.next_sibling
    info = n.get_text(strip=True)
    info_list.append(info)

print(info_list)
```

```python
# 영화 감독 / 출연진
info_list2 = []
items = soup.select("section.wide_dark_bg_wrap ul.info_list a")
for i in items:
    info = i.get_text(strip=True)
    info_list2.append(info)

print(info_list2)
```

```python
# 영화 아이디 리스트
movie_id_list = [63091, 62585, 63075, 62480, 63032, 62799, 63282, 62818, 63108, 63240]
```

```python
info_list = []
temp_list = []
info_list2 = []
```

```python
for d in movie_id_list:
    url = f'https://cine21.com/movie/info/?movie_id={d}'
    response = requests.get(url, headers = header)
    html = response.text
    soup = BeautifulSoup(html, "html.parser")
    items = soup.select("section.wide_dark_bg_wrap ul.info_list li p.title")
    items2 = soup.select("section.wide_dark_bg_wrap ul.info_list a")
    for i in items:
        n = i.next_sibling
        info = n.get_text(strip=True)
        info_list.append(info)
    for i in items2:
        info2 = i.get_text(strip=True)
        info_list2.append(info2)
```

```python
for d in movie_id_list:
    url = f'https://cine21.com/movie/info/?movie_id={d}'
    response = requests.get(url, headers = header)
    html = response.text
    soup = BeautifulSoup(html, "html.parser")
    items = soup.select("section.wide_dark_bg_wrap ul.info_list li p.title")
    for i in items:
        n = i.next_sibling
        info = n.get_text(strip=True)
        temp_list = []
        temp_list.append(info)
        info_list.append(temp_list)


print(info_list)
```
