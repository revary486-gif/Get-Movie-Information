```python
movie_links = []

item = soup.select('section[style^="/height"]')

for link in item:
    href = link.get("href")
    movie_links.append(href)

print(movie_links)
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import (TimeoutException, UnexpectedAlertPresentException)
import time 

link_lists = []
driver = webdriver.Chrome()

driver.get("https://cine21.com/movie/rank/?idx=1")

while True:
    try:
        more_button = WebDriverWait(driver, 3).until(
            EC.element_to_be_clickable(
                (By.CSS_SELECTOR, "button.btn_tertiary.btn_size_l.btn_more")
            )
        )

        driver.execute_script(
            "arguments[0].scrollIntoView();",
            more_button
        )

        more_button.click()

    except UnexpectedAlertPresentException:
        break

html = driver.page_source
soup = BeautifulSoup(html, "html.parser")

items = soup.select('a[href*="movie_id"]')

for item in items:
    mv_link = item.get("href")
    link_lists.append(mv_link)
    print(item.get("href"))

driver.quit()
```
