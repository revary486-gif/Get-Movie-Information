### Q1. 티비랭킹닷컴 (https://workey.codeit.kr/ratings/index) 의 응답 코드를 확인하고 HTML 코드를 rating_page에 저장하세요.

```python
import requests
```

```python
url = "https://workey.codeit.kr/ratings/index"

response = requests.get(url)
rating_page = response.text

print(f"응답 상태: ", response.status_code)
print(f"HTML 코드: ", rating_page[:100])
```

### Q2. 티비랭킹닷컴 (https://workey.codeit.kr/ratings/index) 에서 2010년 1월에 해당하는 HTML 코드를 rating_pages에 저장하세요.

```python
# url = "https://workey.codeit.kr/ratings/index?year=2010&month=1&weekIndex=0"
rating_pages = []
weekIndex = range(0, 5)

for i in weekIndex:
    response = requests.get(f"https://workey.codeit.kr/ratings/index?year=2010&month=1&weekIndex={i}")
    rating_page = response.text
    rating_pages.append(rating_page)

print(len(rating_pages))
print(rating_pages)
```

- 실수했던 부분
1. f-string 쓸 때 i를 중괄호 안에 넣었어야 했는데 weekIndex를 자꾸 넣으려고 해서 error 났었음
2. 드랍박스 상태에 따라 url이 변화한다는 힌트가 없었다면 해결하지 못했을 것 같음
- 호기심 생기는 점
