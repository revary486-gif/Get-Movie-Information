### Q3. 티비랭킹닷컴 (https://workey.codeit.kr/ratings/index) 에서 처음 3년(2010~2012)간의 데이터를 포함한 모든 페이지의 HTML 코드를 rating_pages 리스트에 저장하세요.

```python
# url = "https://workey.codeit.kr/ratings/index?year=2010&month=1&weekIndex=0"

rating_pages = []

years = range(2010, 2013)
months = range(1, 13)
weekindex = range(0, 5)

for i in years:
    for j in months:
        for k in weekindex:
            url = f"https://workey.codeit.kr/ratings/index?year={i}&month={j}&weekIndex={k}"
            response = requests.get(url)
            rating_page = response.text
            rating_pages.append(rating_page)


print(len(rating_pages))
print(rating_pages[-5:])
```

 - 왜 오래걸리나 생각해보니 100개씩 가져오게끔 print문을 작성했었는데 앞서 append를 사용하다보니 각각의 페이지가 하나로 취급되어 100페이지 출력되어 더 오래걸렸던 것 같음

### Q4. 오렌지 보틀 (https://workey.codeit.kr/orangebottle/index) 에서 모든 지점의 전화번호를 태그없이 데이터만 출력하여 phone_numbers 리스트에 저장하세요.

```python
from bs4 import BeautifulSoup
```

```python
url = "https://workey.codeit.kr/orangebottle/index"
phone_numbers = []

response = requests.get(url)
html = response.text
soup = BeautifulSoup(html, "html.parser")
item = soup.select("span.phoneNum")

for i in item:
    phone_num = i.get_text()
    phone_numbers.append(phone_num)

print(len(phone_numbers))
print(phone_numbers)
```

### Q5. 오렌지 보틀 (https://workey.codeit.kr/orangebottle/index) 의 모든 지점의 이름, 주소, 전화번호를 '오렌지_보틀.xlsx'에 저장하세요. 헤더 행은 '지점 이름', '주소', '전화번호'로 저장하세요.

```python
from openpyxl import Workbook
```

```python
url = "https://workey.codeit.kr/orangebottle/index"
branch_names = []
branch_address = []
phone_numbers = []

response = requests.get(url)
html = response.text
soup = BeautifulSoup(html, "html.parser")
items_1 = soup.select("p.city")
items_2 = soup.select("p.address")
items_3 = soup.select("span.phoneNum")

for i in items_1:
    branch_name = i.get_text(strip=True)
    branch_names.append(branch_name)
for j in items_2:
    address = j.get_text(strip=True)
    branch_address.append(address)
for k in items_3:
    phone_num = k.get_text()
    phone_numbers.append(phone_num)
```

```python
wb = Workbook(write_only=True)
ws = wb.create_sheet()
ws.append(['지점_이름', '주소', '전화번호'])

for row in zip(branch_names, branch_address, phone_numbers):
    ws.append(row)

wb.save('branch_details.xlsx')
```
